**REUSABLE ATTRITION ANALYSIS FUNCTION**

```python
def attrition_analysis(data, column):
    result = (
        data.groupby(column, observed=True)
            .agg(
                Employees=("EmployeeNumber", "count"),
                Attrition_Count=(
                    "Attrition",
                    lambda x: (x == "Yes").sum()
                )
            )
            .reset_index()
    )

    result["Attrition_Rate_%"] = (
        result["Attrition_Count"]
        / result["Employees"]
        * 100
    ).round(2)

    return result.sort_values(
        "Attrition_Rate_%",
        ascending=False
    )
```

```python
attrition_analysis(df, "Department")
```

```python
attrition_analysis(df, "JobRole")
```

```python
attrition_analysis(df, "OverTime")
```

**CHI-SQUARE FUNCTION**

```python
def chi_square_test(data, variable, target="Attrition"):
    table = pd.crosstab(
        data[variable],
        data[target]
    )

    chi2, p_value, dof, expected = stats.chi2_contingency(table)

    result = {
        "Variable": variable,
        "Chi-Square": round(chi2, 4),
        "p-value": round(p_value, 6),
        "Degrees of Freedom": dof,
        "Significant_At_5%": p_value < 0.05
    }

    return result
```

**EFFECT SIZE**

```python
def cramers_v(data, variable, target="Attrition"):
    table = pd.crosstab(data[variable], data[target])

    chi2 = stats.chi2_contingency(table)[0]
    n = table.sum().sum()

    r, k = table.shape

    phi2 = chi2 / n

    correction = ((k - 1) * (r - 1)) / (n - 1)

    phi2_corrected = max(
        0,
        phi2 - correction
    )

    r_corrected = r - ((r - 1) ** 2 / (n - 1))
    k_corrected = k - ((k - 1) ** 2 / (n - 1))

    denominator = min(
        k_corrected - 1,
        r_corrected - 1
    )

    return np.sqrt(phi2_corrected / denominator)
```

```python
cramers_v(df, "OverTime")
```

**EFFECT SIZE FOR ALL VARIABLE**

```python
effect_results = []

for column in categorical_tests:
    effect_results.append({
        "Variable": column,
        "Cramers_V": round(
            cramers_v(df, column),
            4
        )
    })

effect_results = pd.DataFrame(effect_results)

effect_results = effect_results.sort_values(
    "Cramers_V",
    ascending=False
)

effect_results
```

```python
statistical_summary = chi_results.merge(
    effect_results,
    on="Variable"
)

statistical_summary = statistical_summary.sort_values(
    "Cramers_V",
    ascending=False
)

statistical_summary
```

**Interpret Cramér's V**

```python
def interpret_effect(v):
    if v < 0.10:
        return "Very Weak"
    elif v < 0.20:
        return "Weak"
    elif v < 0.30:
        return "Moderate"
    else:
        return "Relatively Strong"

statistical_summary["Effect_Interpretation"] = (
    statistical_summary["Cramers_V"]
    .apply(interpret_effect)
)

statistical_summary
```
